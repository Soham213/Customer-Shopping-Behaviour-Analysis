```python
df = df.drop_duplicates()
df = df[df['Customer ID'] <= 3900].copy()
```

```python
for col in [ 'Review Rating']:
    df[col] = df[col].fillna(df.groupby('Category')[col].transform('median'))
##just add the Purchase Amount (USD) and the Previous Purchases along side the review rating above ,into single inverted commas 
##and seperate them with comma
##df.loc[df['Category'] == 'Electronics', 'Size'] = df.loc[df['Category'] == 'Electronics', 'Size'].fillna('N/A')
##df['Size'] = df['Size'].fillna(df['Size'].mode()[0])
```

```python
df.columns
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a new column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df['season'] = df['season'].replace({'Autumn': 'Fall'})
```

```python
# Map the 4 city names in 'location' to their respective US states
city_to_state = {
    'Los Angeles': 'California',
    'Houston': 'Texas',
    'Phoenix': 'Arizona',
    'Chicago': 'Illinois'
}

df['location'] = df['location'].replace(city_to_state)

# Verify: unique count should drop from 54 to exactly 50 US states
print("Unique locations:", df['location'].nunique())
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus
import pandas as pd

host = r"localhost\SQLEXPRESS"
database = "customer_behaviour_analysis"
driver = quote_plus("ODBC Driver 17 for SQL Server")

engine = create_engine(f"mssql+pyodbc://{host}/{database}?driver={driver}&Trusted_Connection=yes")

# Write cleaned DataFrame to the 'CustomerDataset' table in SQL Server
df.to_sql("CustomerDataset", engine, if_exists="replace", index=False)

# Verify the table was updated
print(pd.read_sql("SELECT TOP 5 * FROM CustomerDataset;", engine))
```
